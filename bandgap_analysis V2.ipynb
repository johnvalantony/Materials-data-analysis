```python
import pandas as pd
import matplotlib.pyplot as plt

# Real materials data
data = {
    "Material": [
        "Cobalt Tungstate", "Silicon", "Titanium Dioxide",
        "Zinc Oxide", "Gallium Nitride", "Copper Oxide",
        "Iron Oxide", "Cadmium Sulfide", "Zinc Sulfide", "Tin Oxide"
    ],
    "Bandgap_eV": [2.8, 1.1, 3.2, 3.4, 3.4, 1.2, 2.2, 2.4, 3.6, 3.6],
    "Type": [
        "Semiconductor", "Semiconductor", "Insulator",
        "Semiconductor", "Semiconductor", "Semiconductor",
        "Semiconductor", "Semiconductor", "Insulator", "Semiconductor"
    ],
    "Application": [
        "Photocatalysis", "Solar cells", "Photocatalysis",
        "LEDs", "LEDs", "Solar cells",
        "Batteries", "Solar cells", "Phosphors", "Gas sensors"
    ]
}

df = pd.DataFrame(data)
print(df)
print("\nAverage Bandgap:", df["Bandgap_eV"].mean(), "eV")
print("Highest Bandgap:", df.loc[df["Bandgap_eV"].idxmax(), "Material"])
print("Lowest Bandgap:", df.loc[df["Bandgap_eV"].idxmin(), "Material"])
```

```python
from matplotlib.patches import Patch

# Sort by bandgap
df = df.sort_values("Bandgap_eV")

# Colors by type
colors = {"Semiconductor": "steelblue", "Insulator": "coral"}
bar_colors = [colors[t] for t in df["Type"]]

plt.figure(figsize=(12, 6))
bars = plt.bar(df["Material"], df["Bandgap_eV"], color=bar_colors)

plt.title("Bandgap Analysis of 10 Key Materials", fontsize=14)
plt.xlabel("Material")
plt.ylabel("Bandgap (eV)")
plt.xticks(rotation=45, ha='right')

# Value labels on each bar
for bar, val in zip(bars, df["Bandgap_eV"]):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.05,
             f"{val} eV", ha='center', va='bottom', fontsize=9)

# Legend
legend = [Patch(color='steelblue', label='Semiconductor'),
          Patch(color='coral', label='Insulator')]
plt.legend(handles=legend)

plt.tight_layout()
plt.savefig("bandgap_analysis_v2.png")
plt.show()

from google.colab import files
files.download("bandgap_analysis_v2.png")
```
