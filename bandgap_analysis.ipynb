```python
# My first data analysis - John Valan Tony
materials = ["Cobalt Tungstate", "Silicon", "Titanium Dioxide", "Zinc Oxide"]
bandgaps = [2.8, 1.1, 3.2, 3.4]  # in electron volts (eV)

for i in range(len(materials)):
    print(f"{materials[i]} has a bandgap of {bandgaps[i]} eV")
```

```python
import matplotlib.pyplot as plt

materials = ["Cobalt Tungstate", "Silicon", "Titanium Dioxide", "Zinc Oxide"]
bandgaps = [2.8, 1.1, 3.2, 3.4]

plt.figure(figsize=(8, 5))
plt.bar(materials, bandgaps, color=['steelblue', 'orange', 'green', 'red'])
plt.title("Bandgap Comparison of Materials")
plt.xlabel("Material")
plt.ylabel("Bandgap (eV)")
plt.tight_layout()
plt.savefig("bandgap_chart.png")
plt.show()

print("Chart saved!")
```
